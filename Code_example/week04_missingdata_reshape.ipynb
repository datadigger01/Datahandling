##### 2.2 셀수 있는(countable) 변수 와 범주형 변수에 대해 value_counts(dropna=False) 함수를 이용하여 빈도수 체크(missing 값 포함)

```python
# prompt: value_counts() 함수를 이용하여 'Age_Grp', 'Gender', 'Apply_CH', 're_hire_freq_grp',  'Q2', 'Q7', 'Q8', 'Q18', 'Q22', 'Q38',  'rank_INDEX14', 'rank_INDEX18','rank_INDEX23', 'rank_INDEX27', 'rank_INDEX28', 'rank_INDEX29','rank_INDEX32', 'rank_INDEX44', 'rank_INDEX93', 'rank_AVG_AMT_6M'' 컬럼에 대해 결과를 산출할 수 있느 코드를 제공해주세요

for column in ['Age_Grp', 'Gender', 'Apply_CH', 're_hire_freq_grp', 'Q2', 'Q7', 'Q8', 'Q18', 'Q22', 'Q38', 'rank_INDEX14', 'rank_INDEX18', 'rank_INDEX23', 'rank_INDEX27', 'rank_INDEX28', 'rank_INDEX29', 'rank_INDEX32', 'rank_INDEX44', 'rank_INDEX93', 'rank_AVG_AMT_6M']:
  print(f"Value counts for {column}:")
  print(df[column].value_counts(dropna=False))
  print("\n")
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

columns_to_plot = [
    'rank_INDEX14', 'rank_INDEX18', 'rank_INDEX23',
    'rank_INDEX27', 'rank_INDEX28', 'rank_INDEX29',
    'rank_INDEX32', 'rank_INDEX44', 'rank_INDEX93'
]

# Calculate number of rows and columns for the subplot grid
n_cols = 3
n_rows = (len(columns_to_plot) + n_cols - 1) // n_cols # Ceiling division

fig, axes = plt.subplots(n_rows, n_cols, figsize=(n_cols * 5, n_rows * 4))
axes = axes.flatten() # Flatten the 2D array of axes for easy iteration

for i, col in enumerate(columns_to_plot):
    if i < len(axes): # Ensure we don't go out of bounds if there are more columns than subplots
        sns.countplot(x=col, data=df, ax=axes[i], palette='viridis', hue=col, legend=False)
        axes[i].set_title(f'Frequency of {col}')
        axes[i].set_xlabel(col)
        axes[i].set_ylabel('Count')

# Hide any unused subplots
for j in range(len(columns_to_plot), len(axes)):
    fig.delaxes(axes[j])

plt.tight_layout()
plt.show()
```

##### 2.3 실수타입으로 지정된 rank_INDEX? 컬럼값을 최빈값으로 대체

```python
columns_to_process = ['rank_INDEX14', 'rank_INDEX18', 'rank_INDEX23', 'rank_INDEX27', 'rank_INDEX28',
                      'rank_INDEX29', 'rank_INDEX32', 'rank_INDEX44', 'rank_INDEX93', 'rank_AVG_AMT_6M']

for column in columns_to_process:
  # 최빈값 찾기
  most_frequent_value = df[column].mode()[0]
  # NaN 값을 최빈값으로 대체하고 정수형으로 변환
  df[column] = df[column].fillna(most_frequent_value).astype(int)

# 결과 확인
df.info()
```

```python
# Gender 컬럼을 문자타입으로 변환
df['Gender'] = df['Gender'].astype(str)

df.info()
```

```python
# 변경된 데이터프레임 확인
print(df.isnull().sum())
```
