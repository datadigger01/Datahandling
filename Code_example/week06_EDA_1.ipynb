## 데이터 불러오기
++ WestRoxBury.csv 파일을 pandas로 불러오기
#### 컬럼설명
 -  TOTAL VALUE : 주택가격(단위: 1000달러)
 -  TAX : 세금, 주택가격에 세율을 곱한 값에 근거한 세금계산서 금액(단위: 달러)
 -  LOT SQFT : 총 부지 면적(단위: 제곱 피트)
 -  YR BUILT : 건축 연도
 -  GROSS AREA : 총 바닥면적
 -  LIVING AREA : 주거 공간 총 면적(단위: 제곱 피트)
 -  FLOORS : 층 수
 -  ROOMS : 총 방 수
 -  BEDROOMS : 총 침실 수
 -  FULL BATH : 총 욕실 개수
 -  HALF BATH : 총 보조욕실 개수
 -  KITCHEN : 총 주방 개수
 -  FRIREPLACE : 총 벽난로 개수
 -  REMODEL : 언제 리모델링을 했는지 (최근/오래전/안함)

```python
df = pd.read_csv('https://github.com/datadigger01/Datahandling/raw/main/Data/WestRoxbury.csv')
df
```

```python
print(df.shape)
print(df.columns)
print(df.info())
```

### 컬럼명 일괄 바꾸기

```python
# prompt: 컬럼명을 바꾸는 코드를 알려주세요

new_column_names = {
    'TOTAL VALUE ': 'TOTAL_VALUE',
    'TAX ': 'TAX',
    'LOT SQFT ': 'LOT_SQFT',
    'YR BUILT': 'YR_BUILT',
    'GROSS AREA ': 'GROSS_AREA',
    'LIVING AREA': 'LIVING_AREA',
    'FLOORS ': 'FLOORS',
    'ROOMS ': 'ROOMS',
    'BEDROOMS ': 'BEDROOMS',
    'FULL BATH': 'FULL_BATH',
    'HALF BATH': 'HALF_BATH',
    'KITCHEN ': 'KITCHEN',
    'FIREPLACE ': 'FIREPLACE',
    'REMODEL ': 'REMODEL'
}

df = df.rename(columns=new_column_names)
print(df.columns)
df
```

### 데이터 정제(Cleansing) - missing 값 처리(대체)

```python
print(df.isna().sum())  # 컬럼(변수)별 missing 발생건수
print("\n",df[['REMODEL']].value_counts())
df['REMODEL'].value_counts(dropna=False, normalize=True)  # dropna 옵션은 missing값을 포함하는나 제외해서 하는냐에 대한 옵션
```

#### REMODEL 컬럼 결측값을 'None' Category 값으로 대체

```python
# prompt: df의 REMODEL 컬럼 결측값을 'None'으로 대체
df['REMODEL'] = df['REMODEL'].fillna('None')

# 결측치 대체 결과 및 빈도수 확인
print(df['REMODEL'].value_counts(dropna=False))
display(df.head())
print(df.info())
```

## 분포(Distribution) - 예) TOTAL_VALUE 컬럼에 대한 분포확인

```python
df['TOTAL_VALUE'].describe()
```

```python
# prompt: give me the code to show TOTAL_VALUE distribution

# 히스토그램으로 TOTAL_VALUE 분포 확인
plt.figure(figsize=(10, 6))
sns.histplot(df['TOTAL_VALUE'], bins=30, kde=True)
plt.title('Distribution of TOTAL_VALUE')
plt.xlabel('TOTAL_VALUE')
plt.ylabel('Frequency')
plt.show()
```

```python
from scipy.stats import skew, kurtosis
print("왜도(Skewness):", skew(df['TOTAL_VALUE']))
print("첨도(Kurtosis):", kurtosis(df['TOTAL_VALUE']))
print("평균(Mean):", df['TOTAL_VALUE'].mean())
print("중앙값(Median):", df['TOTAL_VALUE'].median())
print("최빈값(Mode):", df['TOTAL_VALUE'].mode()[0])
#print("분산(Variance):", df['TOTAL_VALUE'].var())
print("표준편차(Standard Deviation):", df['TOTAL_VALUE'].std())
```

## 그래프 (hist) - 예) 건축년도별 건축건물 건수

```python
yr_built_counts = df['YR_BUILT'].value_counts()

# 막대 그래프로 시각화
plt.figure(figsize=(12, 6))
plt.bar(yr_built_counts.index, yr_built_counts.values)
plt.xlabel('YR_BUILT')
plt.ylabel('Count')
plt.title('YR_BUILT Value Count')
plt.xticks(rotation=90)  # x축 레이블 회전
plt.show()
```

```python
# prompt: please, give me the code to show the plot which YR_BUILT value has more than 1900

# YR_BUILT가 1900보다 큰 데이터만 선택
df_filtered = df[df['YR_BUILT'] > 1900]

# YR_BUILT의 값별 빈도수 계산
yr_built_counts = df_filtered['YR_BUILT'].value_counts()

# 막대 그래프로 시각화
plt.figure(figsize=(12, 6))
plt.bar(yr_built_counts.index, yr_built_counts.values)
plt.xlabel('YR_BUILT')
plt.ylabel('Count')
plt.title('YR_BUILT Value Count (Greater than 1900)')
plt.xticks(rotation=90)  # x축 레이블 회전
plt.show()
```

##### 정규화(Normalization) - 연속형 형태의 변수를 [0,1] 사이의 값으로 변환

Normalization.png

```python
# prompt: scikit learn에서 제공하는 정규화 함수를 이용하여 연속형 변수를 정규화하되 컬럼명이 다른 이름을 갖도록 하는 코드를 제공해주세요

from sklearn.preprocessing import MinMaxScaler, StandardScaler

# 정규화할 컬럼 선택 (연속형 변수만 선택)
numerical_cols = ['LOT_SQFT', 'GROSS_AREA', 'LIVING_AREA']

# MinMaxScaler 객체 생성
scaler = MinMaxScaler()

# 선택한 컬럼에 대해 정규화 수행
df_normalized = df.copy()  # 원본 데이터프레임 복사
df_normalized[numerical_cols] = scaler.fit_transform(df[numerical_cols])

# 컬럼명 변경 (새로운 컬럼명 지정)
new_column_names = [col + '_normalized' for col in numerical_cols]
df_normalized.rename(columns=dict(zip(numerical_cols, new_column_names)), inplace=True)
```

```python
# prompt: give me the code to show TOTAL_VALUE distribution

# 히스토그램으로 TOTAL_VALUE 분포 확인
plt.figure(figsize=(10, 6))
sns.histplot(df['LIVING_AREA'], bins=20, kde=True)
plt.title('Distribution of LIVING_AREA')
plt.xlabel('LIVING_AREA')
plt.ylabel('Frequency')
plt.show()


plt.figure(figsize=(10, 6))
sns.histplot(df_normalized['LIVING_AREA_normalized'], bins=20, kde=True)
plt.title('Distribution of LIVING_AREA_normalized')
plt.xlabel('LIVING_AREA_normalized')
plt.ylabel('Frequency')
plt.show()
```
